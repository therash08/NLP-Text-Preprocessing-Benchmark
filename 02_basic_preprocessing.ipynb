```python

# =====================================================
# NLP PROJECT
# PHASE 2: BASIC TEXT PREPROCESSING
# STEP 1: ENVIRONMENT & DATASET SETUP
# =====================================================

import sys
import pandas as pd
import numpy as np
import nltk

print("=" * 60)
print("PHASE 2: BASIC TEXT PREPROCESSING")
print("=" * 60)

# Verify Python environment
print("\nPython Environment:")
print(sys.executable)

# Load datasets
train_df = pd.read_csv("sentiment_train.csv")
validation_df = pd.read_csv("sentiment_validation.csv")
test_df = pd.read_csv("sentiment_test.csv")

# Store datasets
datasets = {
    "Train": train_df,
    "Validation": validation_df,
    "Test": test_df
}

# Verify dataset loading
for name, df in datasets.items():

    print(f"\n{name} Dataset")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

print("\nAll datasets loaded successfully!")

# Show original text samples
print("\nOriginal Text Samples:")

display(
    train_df[["text", "label"]].head(5)
)
```
